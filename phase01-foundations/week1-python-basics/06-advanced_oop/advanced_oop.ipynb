## Part 1: Inheritance - The Basics

### What is Inheritance?
Inheritance allows a class to inherit attributes and methods from another class.

Real-world analogy:
- Parent class (Animal): Has general characteristics (breathe, eat, sleep)
- Child class (Dog): Inherits those + adds specific ones (bark, fetch)

```python
# Parent class (Base class / Superclass)
class Animal:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        print(f"{self.name} is eating")
    
    def sleep(self):
        print(f"{self.name} is sleeping")

# Child class (Derived class / Subclass)
class Dog(Animal):  # Dog inherits from Animal
    def bark(self):
        print(f"{self.name} says: Woof!")

# Create instances
animal = Animal("Generic Animal")
dog = Dog("Buddy")

# Dog has access to Animal methods
dog.eat()    # Buddy is eating (inherited)
dog.sleep()  # Buddy is sleeping (inherited)
dog.bark()   # Buddy says: Woof! (own method)

# Animal doesn't have Dog methods
# animal.bark()  # AttributeError!
```

### Why use Inheritance?

```python
# WITHOUT inheritance (code duplication)
class Dog:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        print(f"{self.name} is eating")
    
    def sleep(self):
        print(f"{self.name} is sleeping")
    
    def bark(self):
        print(f"{self.name} barks")

class Cat:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        print(f"{self.name} is eating")  # DUPLICATED!
    
    def sleep(self):
        print(f"{self.name} is sleeping")  # DUPLICATED!
    
    def meow(self):
        print(f"{self.name} meows")

# WITH inheritance (DRY - Don't Repeat Yourself)
class Animal:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        print(f"{self.name} is eating")
    
    def sleep(self):
        print(f"{self.name} is sleeping")

class Dog(Animal):
    def bark(self):
        print(f"{self.name} barks")

class Cat(Animal):
    def meow(self):
        print(f"{self.name} meows")
```

### Checking Inheritance

```python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

# Check if object is instance of a class
print(isinstance(dog, Dog))     # True
print(isinstance(dog, Animal))  # True (Dog IS an Animal)
print(isinstance(dog, object))  # True (everything inherits from object)

# Check if class is subclass of another
print(issubclass(Dog, Animal))  # True
print(issubclass(Animal, Dog))  # False
```

## Part 2: Method Overriding and super()

### Method Overriding 
Override a parent method to customize it in the child class.

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def make_sound(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    # Override the make_sound method
    def make_sound(self):
        print(f"{self.name} barks: Woof!")

class Cat(Animal):
    # Override the make_sound method
    def make_sound(self):
        print(f"{self.name} meows: Meow!")

# Same method name, different behavior
animal = Animal("Generic")
dog = Dog("Buddy")
cat = Cat("Whiskers")

animal.make_sound()  # Generic makes a sound
dog.make_sound()     # Buddy barks: Woof!
cat.make_sound()     # Whiskers meows: Meow!
```

### Using super() - Extending Parent Methods
super() calls parent class methood, allowing you to extend it rather than completely replace it. 

```python
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
    
    def describe(self):
        return f"{self.name} is a {self.species}"

class Dog(Animal):
    def __init__(self, name, breed):
        # Call parent __init__ using super()
        super().__init__(name, "Dog")
        # Add additional attribute
        self.breed = breed
    
    def describe(self):
        # Call parent describe() and extend it
        parent_description = super().describe()
        return f"{parent_description}, breed: {self.breed}"

dog = Dog("Buddy", "Golden Retriever")
print(dog.describe())
# Buddy is a Dog, breed: Golden Retriever
```

### Common Pattern: Extending init

```python
class Vehicle:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year
    
    def get_info(self):
        return f"{self.year} {self.brand}"

class Car(Vehicle):
    def __init__(self, brand, year, num_doors):
        # Call parent constructor
        super().__init__(brand, year)
        # Add car-specific attributes
        self.num_doors = num_doors
    
    def get_info(self):
        # Extend parent method
        base_info = super().get_info()
        return f"{base_info} with {self.num_doors} doors"

car = Car("Toyota", 2024, 4)
print(car.get_info())  # 2024 Toyota with 4 doors
```

### Multiple Levels of Inheritance

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

tesla = ElectricCar("Tesla", "Model 3", "75 kWh")
print(tesla.brand)          # Tesla
print(tesla.model)          # Model 3
print(tesla.battery_size)   # 75 kWh
```

## Part 3: Polymorphism

### What is poylmorphism?
Polymorphism means "many forms" - same interface, different implementations.

```python
class Shape:
    def area(self):
        pass
    
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius

# Polymorphism in action
shapes = [
    Rectangle(10, 5),
    Circle(7),
    Rectangle(3, 8)
]

# Same method call, different behavior
for shape in shapes:
    print(f"Area: {shape.area():.2f}")

# Output:
# Area: 50.00
# Area: 153.94
# Area: 24.00
```

### Polymorphism with Data Sources (Data Engineering Example)

```python
class DataSource:
    """Base class for data sources"""
    def connect(self):
        pass
    
    def read_data(self):
        pass
    
    def close(self):
        pass

class CSVDataSource(DataSource):
    def __init__(self, filepath):
        self.filepath = filepath
    
    def connect(self):
        print(f"Opening CSV file: {self.filepath}")
    
    def read_data(self):
        print(f"Reading data from {self.filepath}")
        return [["Alice", 25], ["Bob", 30]]
    
    def close(self):
        print(f"Closing {self.filepath}")

class DatabaseSource(DataSource):
    def __init__(self, connection_string):
        self.connection_string = connection_string
    
    def connect(self):
        print(f"Connecting to database: {self.connection_string}")
    
    def read_data(self):
        print("Executing query...")
        return [["Charlie", 35], ["Diana", 28]]
    
    def close(self):
        print("Closing database connection")

class APIDataSource(DataSource):
    def __init__(self, url):
        self.url = url
    
    def connect(self):
        print(f"Connecting to API: {self.url}")
    
    def read_data(self):
        print("Fetching data from API...")
        return [["Eve", 32], ["Frank", 29]]
    
    def close(self):
        print("Closing API connection")

# Polymorphic function - works with any DataSource
def process_data_source(source: DataSource):
    source.connect()
    data = source.read_data()
    print(f"Processed {len(data)} records")
    source.close()
    print()

# Same function, different sources
sources = [
    CSVDataSource("data.csv"),
    DatabaseSource("postgresql://localhost/mydb"),
    APIDataSource("https://api.example.com/data")
]

for source in sources:
    process_data_source(source)
```

## Part 4: Multiple Inheritance

### What is Multiple Inheritance?
A class can inherit from multiple parent classes.

```python
class Flyable:
    def fly(self):
        print(f"{self.name} is flying")

class Swimmable:
    def swim(self):
        print(f"{self.name} is swimming")

# Duck inherits from both Flyable and Swimmable
class Duck(Flyable, Swimmable):
    def __init__(self, name):
        self.name = name
    
    def quack(self):
        print(f"{self.name} says: Quack!")

duck = Duck("Donald")
duck.fly()    # Donald is flying
duck.swim()   # Donald is swimming
duck.quack()  # Donald says: Quack!
```

### Method Resolution Order (MRO)
When multiple parents have the same order, Python uses MRO to decide which to call.

```python
class A:
    def process(self):
        print("A's process")

class B:
    def process(self):
        print("B's process")

class C(A, B):  # Inherits from both A and B
    pass

c = C()
c.process()  # Which process() is called?
# Output: A's process (A comes first in inheritance list)

# Check the Method Resolution Order
print(C.mro())
# [<class 'C'>, <class 'A'>, <class 'B'>, <class 'object'>]
```

### Mixin Pattern
Mixins are classes designed to add funtionality to other classes through multiple inheritance.

```python
class JSONMixin:
    """Mixin to add JSON export capability"""
    def to_json(self):
        import json
        return json.dumps(self.__dict__)

class CSVMixin:
    """Mixin to add CSV export capability"""
    def to_csv(self):
        return ", ".join(str(v) for v in self.__dict__.values())

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Add export capabilities using mixins
class ExportablePerson(Person, JSONMixin, CSVMixin):
    pass

person = ExportablePerson("Alice", 25)
print(person.to_json())  # {"name": "Alice", "age": 25}
print(person.to_csv())   # Alice, 25
```

## Part 5: Abstract Classes
Abstract classes are templates that cannot be instantiated directly. They define methods that must be implemented by child classes.

```python
from abc import ABC, abstractmethod

class Animal(ABC):  # Inherit from ABC (Abstract Base Class)
    def __init__(self, name):
        self.name = name
    
    @abstractmethod
    def make_sound(self):
        """This method MUST be implemented by child classes"""
        pass
    
    @abstractmethod
    def move(self):
        """This method MUST be implemented by child classes"""
        pass
    
    def eat(self):
        """Concrete method - optional to override"""
        print(f"{self.name} is eating")

# This will cause an error:
# animal = Animal("Generic")  # TypeError: Can't instantiate abstract class

class Dog(Animal):
    def make_sound(self):
        return "Woof!"
    
    def move(self):
        return "Running on four legs"

# Now we can create instances
dog = Dog("Buddy")
print(dog.make_sound())  # Woof!
print(dog.move())        # Running on four legs
dog.eat()                # Buddy is eating
```

### Real-World Example: Data Processor

```python
from abc import ABC, abstractmethod

class DataProcessor(ABC):
    """Abstract base class for all data processors"""
    
    @abstractmethod
    def load_data(self, source):
        """Load data from source"""
        pass
    
    @abstractmethod
    def transform_data(self, data):
        """Transform the data"""
        pass
    
    @abstractmethod
    def save_data(self, data, destination):
        """Save processed data"""
        pass
    
    def process(self, source, destination):
        """Template method - defines the process flow"""
        print("Starting data processing...")
        data = self.load_data(source)
        transformed = self.transform_data(data)
        self.save_data(transformed, destination)
        print("Data processing complete!")

class CSVProcessor(DataProcessor):
    def load_data(self, source):
        print(f"Loading data from CSV: {source}")
        return [["Alice", "25"], ["Bob", "30"]]
    
    def transform_data(self, data):
        print("Transforming CSV data...")
        # Convert ages to integers
        return [[row[0], int(row[1])] for row in data]
    
    def save_data(self, data, destination):
        print(f"Saving to CSV: {destination}")
        # Would write to file here

class JSONProcessor(DataProcessor):
    def load_data(self, source):
        print(f"Loading data from JSON: {source}")
        return [{"name": "Charlie", "age": "35"}]
    
    def transform_data(self, data):
        print("Transforming JSON data...")
        # Convert ages to integers
        for item in data:
            item["age"] = int(item["age"])
        return data
    
    def save_data(self, data, destination):
        print(f"Saving to JSON: {destination}")
        # Would write to file here

# Use polymorphically
processors = [
    CSVProcessor(),
    JSONProcessor()
]

for processor in processors:
    processor.process("input.data", "output.data")
    print()
```

## Part 6: Class Methods and Static Methods

### Instance Methods vs Class Methods vs Static Methods

```python
class MyClass:
    class_variable = 0
    
    def __init__(self, value):
        self.instance_variable = value
    
    # Instance method (regular method)
    def instance_method(self):
        """Has access to self (the instance)"""
        return f"Instance value: {self.instance_variable}"
    
    # Class method
    @classmethod
    def class_method(cls):
        """Has access to cls (the class itself)"""
        return f"Class variable: {cls.class_variable}"
    
    # Static method
    @staticmethod
    def static_method(x, y):
        """No access to self or cls - just a regular function"""
        return x + y

# Usage
obj = MyClass(10)

# Instance method - needs an instance
print(obj.instance_method())  # Instance value: 10

# Class method - can be called on class or instance
print(MyClass.class_method())  # Class variable: 0
print(obj.class_method())      # Class variable: 0

# Static method - utility function
print(MyClass.static_method(5, 3))  # 8
print(obj.static_method(5, 3))      # 8
```

### When to Use Each

```python
class Person:
    population = 0  # Class variable
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.population += 1
    
    # Instance method - operates on instance data
    def celebrate_birthday(self):
        self.age += 1
        return f"{self.name} is now {self.age}!"
    
    # Class method - alternative constructor
    @classmethod
    def from_birth_year(cls, name, birth_year):
        """Create Person from birth year instead of age"""
        current_year = 2025
        age = current_year - birth_year
        return cls(name, age)
    
    # Class method - access class data
    @classmethod
    def get_population(cls):
        return f"Total people: {cls.population}"
    
    # Static method - utility function (doesn't need class/instance data)
    @staticmethod
    def is_adult(age):
        return age >= 18

# Regular constructor
person1 = Person("Alice", 25)

# Alternative constructor (class method)
person2 = Person.from_birth_year("Bob", 1990)

# Use static method
print(Person.is_adult(20))  # True
print(person1.is_adult(16))  # False

# Class method
print(Person.get_population())  # Total people: 2
```

### Real-World Example: Database Connection Pool

```python
class DatabaseConnection:
    _connections = []  # Class variable - shared pool
    max_connections = 5
    
    def __init__(self, host, database):
        self.host = host
        self.database = database
        self.connected = False
    
    # Instance method
    def connect(self):
        if len(DatabaseConnection._connections) >= DatabaseConnection.max_connections:
            raise Exception("Connection pool exhausted!")
        self.connected = True
        DatabaseConnection._connections.append(self)
        return f"Connected to {self.database} on {self.host}"
    
    # Class method - factory pattern
    @classmethod
    def create_postgres_connection(cls, database):
        """Create PostgreSQL connection with default settings"""
        return cls("localhost", database)
    
    @classmethod
    def create_mysql_connection(cls, database):
        """Create MySQL connection with default settings"""
        return cls("127.0.0.1", database)
    
    # Class method - manage pool
    @classmethod
    def get_connection_count(cls):
        return len(cls._connections)
    
    @classmethod
    def close_all_connections(cls):
        for conn in cls._connections:
            conn.connected = False
        cls._connections.clear()
        return "All connections closed"
    
    # Static method - utility
    @staticmethod
    def validate_database_name(name):
        """Check if database name is valid"""
        return name.isalnum() and len(name) > 0

# Use factory methods
db1 = DatabaseConnection.create_postgres_connection("mydb")
db2 = DatabaseConnection.create_mysql_connection("testdb")

# Use static method
print(DatabaseConnection.validate_database_name("mydb123"))  # True

# Use class methods
print(DatabaseConnection.get_connection_count())  # 0 (not connected yet)
db1.connect()
print(DatabaseConnection.get_connection_count())  # 1
```

## Part 7: Magic Methods (Dunder Methods)

### What are Magic Methods?
Magic methods (double underscore methods) allow you to define how objects behave with built-in Python operations.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    # String representation
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    
    # Addition
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    # Subtraction
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
    
    # Multiplication (scalar)
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)
    
    # Equality
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    # Length (magnitude)
    def __abs__(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5
    
    # Boolean conversion
    def __bool__(self):
        return self.x != 0 or self.y != 0

# Use magic methods
v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(v1)           # Vector(3, 4) - uses __str__
print(v1 + v2)      # Vector(4, 6) - uses __add__
print(v1 - v2)      # Vector(2, 2) - uses __sub__
print(v1 * 2)       # Vector(6, 8) - uses __mul__
print(v1 == v2)     # False - uses __eq__
print(abs(v1))      # 5.0 - uses __abs__
print(bool(v1))     # True - uses __bool__
```

### Common Magic Methods

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    # String representation
    def __str__(self):
        return f"{self.owner}'s account: ${self.balance}"
    
    def __repr__(self):
        return f"BankAccount('{self.owner}', {self.balance})"
    
    # Comparison operators
    def __eq__(self, other):
        """=="""
        return self.balance == other.balance
    
    def __lt__(self, other):
        """<"""
        return self.balance < other.balance
    
    def __le__(self, other):
        """<="""
        return self.balance <= other.balance
    
    def __gt__(self, other):
        """>"""
        return self.balance > other.balance
    
    def __ge__(self, other):
        """>="""
        return self.balance >= other.balance
    
    # Arithmetic operators
    def __add__(self, amount):
        """account + 100"""
        return BankAccount(self.owner, self.balance + amount)
    
    def __sub__(self, amount):
        """account - 50"""
        return BankAccount(self.owner, self.balance - amount)
    
    # Container emulation
    def __len__(self):
        """len(account) - number of digits in balance"""
        return len(str(int(self.balance)))
    
    def __bool__(self):
        """bool(account) - True if balance > 0"""
        return self.balance > 0
    
    # Context manager (with statement)
    def __enter__(self):
        print(f"Opening {self.owner}'s account")
        return self
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        print(f"Closing {self.owner}'s account")
        return False

# Usage
acc1 = BankAccount("Alice", 1000)
acc2 = BankAccount("Bob", 500)

print(acc1 > acc2)       # True (uses __gt__)
print(acc1 == acc2)      # False (uses __eq__)

acc3 = acc1 + 500        # Uses __add__
print(acc3)              # Alice's account: $1500

# Context manager
with BankAccount("Charlie", 2000) as account:
    print(account)
# Opening Charlie's account
# Charlie's account: $2000
# Closing Charlie's account
```
