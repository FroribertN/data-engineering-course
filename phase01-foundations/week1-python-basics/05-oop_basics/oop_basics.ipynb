## Part 1: Understanding Classes and Objects

### What is a Class?
A class is a blueprint for creating objects. Think of it like a cookie cutter.

#### Real-world analogy:

- __Class__: Blueprint for a house
- __Object__: The actual house built from that blueprint

```python
# Define a class (blueprint)
class Dog:
    pass  # Empty class for now

# Create objects (instances)
dog1 = Dog()
dog2 = Dog()

print(dog1)  # <__main__.Dog object at 0x...>
print(dog2)  # <__main__.Dog object at 0x...> (different address!)

# They are different objects
print(dog1 == dog2)  # False
```

### Adding Attributes(Properties)

```python
class Dog:
    pass

# Create a dog
my_dog = Dog()

# Add attributes dynamically (not recommended)
my_dog.name = "Buddy"
my_dog.age = 3
my_dog.breed = "Golden Retriever"

print(f"{my_dog.name} is a {my_dog.age} year old {my_dog.breed}")
# Buddy is a 3 year old Golden Retriever
```

#### Problem with the above approach:
- Have to manually set attributes for each object
- Easy to forget attributes
- No validation

#### Better approach: Use __init__ constructor

## Part 2: The _ _ init _ _ Constructor

### What is _ _ init _ _?
The _ _ init _ _ method is a special method that runs automatically when you create an object.

```python
class Dog:
    # Constructor - runs when object is created
    def __init__(self, name, age, breed):
        # self refers to the object being created
        self.name = name
        self.age = age
        self.breed = breed

# Create dogs using the constructor
dog1 = Dog("Buddy", 3, "Golden Retriever")
dog2 = Dog("Max", 5, "Labrador")
dog3 = Dog("Luna", 2, "Husky")

# Access attributes
print(dog1.name)   # Buddy
print(dog2.age)    # 5
print(dog3.breed)  # Husky
```

### Understanding __self__
self represents the instance (object) itself.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age    # Instance variable
    
    def describe(self):
        # self lets us access the instance's attributes
        print(f"{self.name} is {self.age} years old")

dog = Dog("Buddy", 3)
dog.describe()  # Buddy is 3 years old

# When you call dog.describe(), Python automatically passes 'dog' as self
# It's like calling: Dog.describe(dog)
```

#### Key points about self:
- Must be the first parameter of instance methods
- Python passes it automatically (you don't write dog.describe(dog))
- You can name it anything, but self is the convention

## Part 3: Instance Methods

### What are methods?
Methods are functions defined inside a class that operate on the object's data.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        """Add money to the account"""
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New balance: ${self.balance}")
        else:
            print("Deposit amount must be positive")
    
    def withdraw(self, amount):
        """Remove money from the account"""
        if amount > self.balance:
            print("Insufficient funds!")
        elif amount <= 0:
            print("Withdrawal amount must be positive")
        else:
            self.balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.balance}")
    
    def get_balance(self):
        """Return current balance"""
        return self.balance

# Create account
account = BankAccount("Alice", 1000)

# Use methods
account.deposit(500)      # Deposited $500. New balance: $1500
account.withdraw(200)     # Withdrew $200. New balance: $1300
account.withdraw(2000)    # Insufficient funds!

print(f"Balance: ${account.get_balance()}")  # Balance: $1300
```

### Methods vs Functions

```python
# Function (standalone)
def deposit_money(account_balance, amount):
    return account_balance + amount

balance = 1000
balance = deposit_money(balance, 500)

# Method (part of class)
class Account:
    def __init__(self, balance):
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount

account = Account(1000)
account.deposit(500)  # Operates on account's own data
```

#### Methods are better because:
- Data and operations are together
- Can't accidentally operate on wrong data
- More organized and maintainable

## Part 4: String Representation

### The Problem

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog = Dog("Buddy", 3)
print(dog)  # <__main__.Dog object at 0x0000015B4FCF4C20>
# Not very helpful!
```

#### Solution: _ _ str _ _ and _ _ repr _ _

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        """String representation for users (readable)"""
        return f"{self.name} (age {self.age})"
    
    def __repr__(self):
        """String representation for developers (unambiguous)"""
        return f"Dog('{self.name}', {self.age})"

dog = Dog("Buddy", 3)

# __str__ is used by print() and str()
print(dog)        # Buddy (age 3)
print(str(dog))   # Buddy (age 3)

# __repr__ is used in the REPL and repr()
dog               # Dog('Buddy', 3)
print(repr(dog))  # Dog('Buddy', 3)

# In lists, __repr__ is used
dogs = [Dog("Buddy", 3), Dog("Max", 5)]
print(dogs)  # [Dog('Buddy', 3), Dog('Max', 5)]
```

#### When to use which:
- _ _ str _ _: Human-readable output for end users
- _ _ repr _ _: Unambiguous representation for debugging
- If you only define one, define _ _ repr _ _

### Good Example

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    
    def __str__(self):
        return f"Account({self.owner}): ${self.balance:.2f}"
    
    def __repr__(self):
        return f"BankAccount('{self.owner}', {self.balance})"

account = BankAccount("Alice", 1500.50)
print(account)        # Account(Alice): $1500.50
print(repr(account))  # BankAccount('Alice', 1500.5)
```

## Part 5: Class Variable vs instance Variable

### Instance Variables
Instance variables are unique to each object.

```python
class Dog:
    def __init__(self, name):
        self.name = name  # Instance variable - different for each dog

dog1 = Dog("Buddy")
dog2 = Dog("Max")

print(dog1.name)  # Buddy
print(dog2.name)  # Max

dog1.name = "Charlie"
print(dog1.name)  # Charlie
print(dog2.name)  # Max (unchanged)
```

### Class Variables
Class variables are shared by all objects of the class.

```python
class Dog:
    # Class variable - shared by all dogs
    species = "Canis familiaris"
    count = 0  # Track number of dogs created
    
    def __init__(self, name):
        self.name = name  # Instance variable
        Dog.count += 1    # Increment class variable

dog1 = Dog("Buddy")
dog2 = Dog("Max")
dog3 = Dog("Luna")

# All dogs share the same species
print(dog1.species)  # Canis familiaris
print(dog2.species)  # Canis familiaris

# Class variable accessed via class name
print(Dog.count)     # 3 (total dogs created)
print(dog1.count)    # 3 (can also access via instance)
```

### When to Use Each

```python
class BankAccount:
    # Class variables (shared)
    bank_name = "Global Bank"
    interest_rate = 0.02
    total_accounts = 0
    
    def __init__(self, owner, balance):
        # Instance variables (unique per account)
        self.owner = owner
        self.balance = balance
        self.account_number = BankAccount.total_accounts + 1
        BankAccount.total_accounts += 1
    
    def apply_interest(self):
        # Use class variable for calculation
        interest = self.balance * BankAccount.interest_rate
        self.balance += interest

account1 = BankAccount("Alice", 1000)
account2 = BankAccount("Bob", 2000)

print(BankAccount.bank_name)      # Global Bank
print(BankAccount.total_accounts) # 2

account1.apply_interest()
print(account1.balance)  # 1020.0
```

#### Use class variables for:
- Constants shared by all objects
- Counters
- Default values
- Configuration

#### Use instance variables for:
- Data unique to each object
- State that changes per object

## Part 6: Encapsulation Basics

### Public vs Private Attributes
In Python, there is no true "private" but we have conventions:

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner           # Public - can be accessed
        self._balance = balance      # Protected - should not be accessed directly
        self.__pin = "1234"          # Private - name mangled
    
    def get_balance(self):
        """Public method to access protected attribute"""
        return self._balance
    
    def verify_pin(self, pin):
        """Public method to check private attribute"""
        return self.__pin == pin

account = BankAccount("Alice", 1000)

# Public attribute
print(account.owner)  # Alice

# Protected attribute (convention: don't access directly)
print(account._balance)  # 1000 (works but discouraged)

# Private attribute (name mangled)
# print(account.__pin)  # AttributeError!

# Access through public methods
print(account.get_balance())    # 1000
print(account.verify_pin("1234"))  # True
```

#### Naming conventions:
- public_var: No underscore, intended to be accessed
- _protected_var: Single underscore, internal use (not enforced)
- __private_var: Double underscore, name mangled (harder to access)

### Using Properties (Getters and Setters)

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # Protected
    
    # Getter
    @property
    def balance(self):
        """Get balance (read-only access)"""
        return self._balance
    
    # Setter
    @balance.setter
    def balance(self, value):
        """Set balance with validation"""
        if value < 0:
            raise ValueError("Balance cannot be negative!")
        self._balance = value

account = BankAccount("Alice", 1000)

# Access like an attribute (but uses getter)
print(account.balance)  # 1000

# Set like an attribute (but uses setter with validation)
account.balance = 1500
print(account.balance)  # 1500

# Validation kicks in
# account.balance = -100  # ValueError: Balance cannot be negative!
```

#### Benefits of properties:
- Control access to attributes
- Add validation when setting values
- Compute values on-the-fly
 Keep interface clean (looks like attribute access)

## Part 7: Building a Complete Class
Building a complete, practical class step by step:

### Example: Task Manager

```python
class Task:
    """Represents a task in a todo list"""
    
    # Class variable
    total_tasks = 0
    
    def __init__(self, title, description="", priority="Medium"):
        """
        Initialize a new task.
        
        Args:
            title: Task title
            description: Detailed description
            priority: High, Medium, or Low
        """
        self.title = title
        self.description = description
        self.priority = priority
        self.completed = False
        self.task_id = Task.total_tasks + 1
        Task.total_tasks += 1
    
    def mark_complete(self):
        """Mark task as completed"""
        self.completed = True
        print(f"Task '{self.title}' marked as complete!")
    
    def mark_incomplete(self):
        """Mark task as incomplete"""
        self.completed = False
        print(f"Task '{self.title}' marked as incomplete")
    
    def update_priority(self, new_priority):
        """Update task priority"""
        valid_priorities = ["High", "Medium", "Low"]
        if new_priority in valid_priorities:
            self.priority = new_priority
            print(f"Priority updated to {new_priority}")
        else:
            print(f"Invalid priority. Choose from: {valid_priorities}")
    
    def __str__(self):
        """User-friendly string representation"""
        status = "Completed" if self.completed else "Incomplete"
        return f"[{status}] {self.title} (Priority: {self.priority})"
    
    def __repr__(self):
        """Developer-friendly representation"""
        return f"Task('{self.title}', '{self.description}', '{self.priority}')"


# Usage
task1 = Task("Learn Python OOP", "Complete 5 challenges", "High")
task2 = Task("Exercise", priority="Medium")
task3 = Task("Read book", "Finish chapter 3", "Low")

print(task1)  # [○] Learn Python OOP (Priority: High)
task1.mark_complete()
print(task1)  # [✓] Learn Python OOP (Priority: High)

print(f"\nTotal tasks created: {Task.total_tasks}")  # 3
```
