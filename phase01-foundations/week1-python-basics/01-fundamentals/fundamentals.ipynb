## Data Types Explained

### 1. Integers (int)

```python
# Whole numbers, positive or negative, no decimal point
age = 25
year = 2026
negative_sum = -100
large_number = 1_000_000  # underscores for readability (same as 1000000)

# You can do math with integers
print(10 + 5)      # 15 - Addition
print(10 - 5)      # 5 - Subtraction
print(10 * 5)      # 50 - Multiplication
print(10 / 5)      # 2.0 - Division (always results/returns in a float)
print(10 // 3)     # 3 - Floor Division (returns the largest integer less than or equal to the division result)
print(10 % 3)      # 1 - Modulus (returns the remainder of the division)
print(10 ** 2)     # 100 - Exponentiation (10 raised to the power of 2)
```

### 2. Floats (float)

```python
# Numbers with decimal points
price = 19.99
temperature = -4.5
pi = 3.14159
scientific_notation = 1.5e3  # 1.5 * 10^3 = 1500.0

# Float precision issues (important to know for data engineering tasks)
print(0.1 + 0.2)  # Expected: 0.3, Actual: 0.30000000000000004
# this is normal in all programming languages due to how computers store decimal numbers
# we will use special libraries later, such as 'decimal' or 'numpy', to handle these cases

# Rounding floats
price = 99.999
print(round(price, 2))  # Rounds to 2 decimal places: 100.0
```

### 3. Strings(str)

```python
# Single or double quotes - both work the same
name = 'Jack'
greeting = "Hello, World!"

# Tripe quotes for multi-line strings
long_text = """
This is a multi-line string.
"""

print(name)
print(long_text)

# String concatenation (joining strings together)
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name # "John Doe"   
print(full_name)

# String repetition
separator = "=" * 50 # "=================================================="
print(separator)

# Escape characters
quote = "He said, \"Hello!\"" # Use backslash (\) for quotes inside strings
quote2 = 'He said, "Hello!"' # Can use single quotes for outside and double inside (vice versa)
path = "C:\\Users\\JohnDoe\\Documents" # Use double backslashes for file paths on Windows
new_line = "First Line\nSecond Line" # \n for new line
tabbed = "Column1\tColumn2\tColumn3" # \t for tab space

print(quote)
print(quote2)
print(path)
print(new_line)
print(tabbed)

# String length
message = "Hello, Data Engineering!"
print(len(message))  # 24 - number of characters in the string

# Important: Strings are immutable - you cannot change them in place
original = "Hello"
print(original)
# original[0] = "h"  # This will raise an error! Cannot change individual characters
# Instead, create a new string
modified = "h" + original[1:]  # "hello"
print(modified)
```

### 4. Booleans (bool)

```python
# True or False values (note the capital T and F)
is_active = True
is_logged_in = False

# Boolean operations
print(True and False)  # False - Logical AND (both must be True to return True)
print(True or False)   # True - Logical OR (at least one must be True to return True)
print(not True)        # False - Logical NOT (inverts the value)

# comparison operators
age = 25
print(age > 18)   # True - Greater than
print(age == 25)  # True - Equal to
print(age != 30)  # True - Not equal to
```

## Type Conversion (Type Casting)

```python
# String to Integer
age_str = "30"
age_int = int(age_str)  # Converts string to integer
print(type(age_str)) # <class 'str'>
print(type(age_int)) # <class 'int'>

# String to Float
price_str = "19.99"
price_float = float(price_str)  # Converts string to float
print(type(price_str)) # <class 'str'>
print(type(price_float)) # <class 'float'>

# Integer/Float to String
age = 25
age_str = str(age)  # Converts integer to string
print(type(age))     # <class 'int'>
print(type(age_str)) # <class 'str'>

# Be careful - not everything can be converted
# int("hello") # This will raise a ValueError because "hello" is not a valid integer
# int("19.99") # This will also raise a ValueError because "19.99" is not a valid integer. Cannot covert float decimal strings directly to int

# Better approach for decimal strings
decimal_str = "25.5"
number = int(float(decimal_str))  # First convert to float, then to int
print(number)  # 25
print(type(decimal_str))  # <class 'str'>
print(type(number))  # <class 'int'>

# Checking types
value = 100
print(type(value))  # <class 'int'>
print(isinstance(value, int))  # True
print(isinstance(value, float))  # False
```

## Operators - Complete Guide

```python
### Arithmetic Operators
a = 10
b = 3

# Basic arithmetic operations
print(a + b)      # 13 - Addition
print(a - b)      # 7 - Subtraction
print(a * b)      # 30 - Multiplication
print(a / b)      # 3.3333333333333335 - Division
print(a // b)     # 3 - Floor Division (rounds down to the nearest whole number, returns an integer)
print(a % b)      # 1 - Modulus (remainder)
print(a ** b)     # 1000 - Exponentiation (a raised to the power of b)

# Compound assignment operators (shortcuts)
x = 10
x += 5  # Equivalent to x = x + 5
print(x)  # 15

x -= 3  # Equivalent to x = x - 3
print(x)  # 12

x *= 2  # Equivalent to x = x * 2
print(x)  # 24

x /= 4  # Equivalent to x = x / 4
print(x)  # 6.0

x //= 2  # Equivalent to x = x // 2
print(x)  # 3.0

x %= 2  # Equivalent to x = x % 2
print(x)  # 1.0

x **= 3  # Equivalent to x = x ** 3
print(x)  # 1.0
```

### Comparison Operators

```python
x = 10
y = 20

print(x == y)  # False - Equal to
print(x != y)  # True - Not equal to
print(x > y)   # False - Greater than
print(x < y)   # True - Less than
print(x >= y)  # False - Greater than or equal to
print(x <= y)  # True - Less than or equal to

# Chaining comparisons (Python-specific feature)
age = 25
print(18 <= age <= 30)  # True - Checks if age is between 18 and 30 inclusive

# Comparing strings (alphabetical order)
print("apple" < "banana")  # True - 'apple' comes before 'banana'
print("grape" > "orange")  # False - 'grape' comes before 'orange'
print("cat" == "cat")      # True - Strings are identical
print("Dog" < "dog")      # True - Uppercase letters come before lowercase letters in ASCII
### Logical Operators
# AND - both conditions must be True
age = 26
has_license = True
can_drive = (age >= 18) and has_license # True if age is 18 or older AND has a license
print(can_drive)  # True

# OR - at least one condition must be True
is_weekend = False
is_holiday = True
can_relax = is_weekend or is_holiday # True if it's weekend OR a holiday
print(can_relax)  # True

# NOT - inverts the boolean value
is_working = False
is_free = not is_working # True if not working
print(is_free)  # True

# Complex conditions (use parentheses for clarity)
age = 25
income = 50000
credit_score = 700

# Loan approval logic:
loan_approved = (age >= 21) and (income > 30000) and (credit_score >= 650)
print(loan_approved)  # True

# Short-circuit evaluation (important concept)
# Python stops evaluating as soon as the result is determined
x = 5
# If first condition is False, second condition is not evaluated
result = (x > 10) and (10 / 0 / 1) # No error! Does not reach division by zero
print(result)  # False
```

## Input and Output - The Details

### Getting User Input

```python
# Input() function always returns a string
name = input("What is your name? ")
print(f"Hello, {name}!")

# For numbers, you MUST convert the string input to int or float
age_str = input("Enter your age: ")
age = int(age_str)  # Convert string to integer
print(f"You are {age} years old.")

# Or do it in one line
height = float(input("Enter your height in meters: "))  # Convert string to float
print(f"You are {height} meters tall.")

# Handling multiple inputs at once
print("Enter two numbers separated by space: ")
num1, num2 = input().split() # Splits input string into two parts
num1 = int(num1)  # Convert first input to integer
num2 = int(num2)  # Convert second input to integer
print(f"Sum: {num1 + num2}")

# Using map() to convert multiple inputs - better approach
num1, num2 = map(int, input("Enter two numbers separated by space: ").split())
print(f"Sum: {num1 + num2}")
```

### Output Formatting - Multiple Methods

```python
name = "Alice"
age = 30
salary = 75000.50

# Method 1 - f-strings (Python 3.6+) Recommended
print(f"My name is {name} and I am {age} years old with a salary of ${salary:.2f} per year.")

# Fomarting numbers in f-strings
price = 1234.56789
print(f'Price: ${price:,.2f}')  # Comma as thousand separator, 2 decimal places
print(f'Price: ${price:.2f}') # 2 decimal places without comma
print(f'Price: ${price: >10.2f}') # Right aligned in a field of width 10
print(f'Price: ${price: <10.2f}') # Left aligned in a

# Expressions inside f-strings
print(f'Next year I will be {age + 1} years old.')
print(f'My monthly salary is ${salary / 12:,.2f}.')

# Method 2 - .format() method (older but still used)
print('My name is {} and I am {} years old.'.format(name, age)) # Using positional arguments
print('My name is {0} and I am {1} years old.'.format(name, age)) # Using indexed placeholders
print('My name is {n} and I am {a} years old.'.format(n=name, a=age)) # Using named placeholders

# Method 3 - % operator (oldest method, not recommended)
print('My name is %s and I am %d years old.' % (name, age)) # %s for string, %d for integer
print('My salary is $%.2f per year.' % salary) # %.2f for float with 2 decimal places

# Method 4 - String concatenation (not recommended for complex strings)
print("My name is " + name + " and I am " + str(age) + " years old.")
```
